```python
import os
import time
import numpy as np
from PIL import Image, ImageEnhance, ImageOps
import torch
import matplotlib.pyplot as plt
from diffusers import (
    StableDiffusionControlNetPipeline,
    StableDiffusionInpaintPipeline,
    ControlNetModel
)
from diffusers.utils import load_image

# Set your GPU
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

# File paths and prompt
original_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/standalone_shirts/1907.jpg"
sketch_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/sketches/1907_sketch.png"
mask_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/masks/1907_mask.png"
prompt = "green jacket"

# Load and preprocess images
original_image = load_image(original_path).resize((512, 512)).convert("RGB")
sketch_image = Image.open(sketch_path).resize((512, 512)).convert("L")
sketch_image = ImageOps.invert(sketch_image)
sketch_image = ImageEnhance.Contrast(sketch_image).enhance(2.0).convert("RGB")

mask_image = Image.open(mask_path).resize((512, 512)).convert("L")
mask_image = mask_image.point(lambda x: 255 if x > 128 else 0).convert("RGB")

# Visual debug
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(original_image); axs[0].set_title("Original"); axs[0].axis("off")
axs[1].imshow(sketch_image); axs[1].set_title("Sketch"); axs[1].axis("off")
axs[2].imshow(mask_image); axs[2].set_title("Mask"); axs[2].axis("off")
plt.tight_layout(); plt.show()

# Load ControlNet and Inpainting models
controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/control_v11p_sd15_scribble", torch_dtype=torch.float16
).to("cuda")

controlnet_pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    controlnet=controlnet,
    torch_dtype=torch.float16
).to("cuda")
controlnet_pipe.safety_checker = None

controlnet_pipe.enable_attention_slicing()

inpaint_pipe = StableDiffusionInpaintPipeline.from_pretrained(
    "runwayml/stable-diffusion-inpainting",
    torch_dtype=torch.float16
).to("cuda")
inpaint_pipe.safety_checker = None

inpaint_pipe.enable_attention_slicing()

# Grid search configs
guidance_scales = list(np.arange(4.0, 10.0, 0.5))  # rows
inference_steps = list(range(20, 50, 5))           # cols

results = []
images = []

# Run grid search
for gs in guidance_scales:
    for steps in inference_steps:
        print(f"🧪 Running: GS={gs}, Steps={steps}")
        start = time.time()

        # Step 1: ControlNet (sketch-guided shirt)
        with torch.inference_mode(), torch.autocast("cuda"):
            control_result = controlnet_pipe(
                prompt=prompt,
                image=sketch_image,
                num_inference_steps=steps,
                guidance_scale=gs
            ).images[0]
        for gs in guidance_scales:
            for steps in inference_steps:
                # Step 2: Inpaint edited shirt onto original background
                with torch.inference_mode(), torch.autocast("cuda"):
                    inpainted_result = inpaint_pipe(
                        prompt=prompt,
                        image=original_image,
                        mask_image=mask_image,
                        num_inference_steps=steps,
                        guidance_scale=gs
                    ).images[0]

        elapsed = time.time() - start
        results.append((gs, steps, elapsed))
        images.append((inpainted_result, gs, steps, elapsed))

# Show all results in a single big grid
cols = len(inference_steps)
rows = len(guidance_scales)
fig, axs = plt.subplots(rows, cols, figsize=(3 * cols, 3 * rows))

for i, (img, gs, steps, t) in enumerate(images):
    row = i // cols
    col = i % cols
    axs[row, col].imshow(img)
    axs[row, col].set_title(f"GS={gs}, Steps={steps}\n{t:.2f}s", fontsize=8)
    axs[row, col].axis("off")

plt.tight_layout()
plt.show()

# Summary
print("\n=== Grid Search Summary (Inference Time) ===")
for gs, steps, t in results:
    print(f"GS={gs}, Steps={steps} → Time={t:.2f}s")

fastest = min(results, key=lambda x: x[2])
print(f"\n✅ Fastest: GS={fastest[0]}, Steps={fastest[1]}, Time={fastest[2]:.2f}s")
```

```python
import os
import time
import numpy as np
from PIL import Image, ImageEnhance, ImageOps
import torch
import matplotlib.pyplot as plt
from diffusers import (
    StableDiffusionControlNetPipeline,
    StableDiffusionInpaintPipeline,
    ControlNetModel
)
from diffusers.utils import load_image

# Set your GPU
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

# File paths and prompt
original_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/standalone_shirts/1907.jpg"
sketch_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/sketches/1907_sketch.png"
mask_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/masks/1907_mask.png"
prompt = "green hoodie"

# Load and preprocess inputs
original_image = load_image(original_path).resize((512, 512)).convert("RGB")

sketch_image = Image.open(sketch_path).resize((512, 512)).convert("L")
sketch_image = ImageOps.invert(sketch_image)
sketch_image = ImageEnhance.Contrast(sketch_image).enhance(2.0).convert("RGB")

mask_image = Image.open(mask_path).resize((512, 512)).convert("L")
mask_image = mask_image.point(lambda x: 255 if x > 128 else 0).convert("RGB")

# === Load models ===
controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/control_v11p_sd15_scribble", torch_dtype=torch.float16
).to("cuda")

controlnet_pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", controlnet=controlnet, torch_dtype=torch.float16
).to("cuda")
controlnet_pipe.safety_checker = None
controlnet_pipe.enable_attention_slicing()

inpaint_pipe = StableDiffusionInpaintPipeline.from_pretrained(
    "runwayml/stable-diffusion-inpainting", torch_dtype=torch.float16
).to("cuda")
inpaint_pipe.safety_checker = None
inpaint_pipe.enable_attention_slicing()

# === Grid settings ===
ctrl_guidance_scales = list(np.arange(4.0, 7.5, 1.5))     # for controlnet
ctrl_steps = list(range(20, 40, 10))

inpaint_guidance_scales = list(np.arange(6.0, 13, 1.5)) # for inpainting
inpaint_steps = list(range(20, 40, 10))

results = []
images = []

# === Grid search ===
for ctrl_gs in ctrl_guidance_scales:
    for ctrl_step in ctrl_steps:
        print(f"\n🎨 ControlNet: GS={ctrl_gs}, Steps={ctrl_step}")
        with torch.inference_mode(), torch.autocast("cuda"):
            controlnet_output = controlnet_pipe(
                prompt=prompt,
                image=sketch_image,
                num_inference_steps=ctrl_step,
                guidance_scale=ctrl_gs
            ).images[0]

        for inpaint_gs in inpaint_guidance_scales:
            for inpaint_step in inpaint_steps:
                print(f"🧩 Inpaint: GS={inpaint_gs}, Steps={inpaint_step}")
                start = time.time()
                with torch.inference_mode(), torch.autocast("cuda"):
                    inpainted_result = inpaint_pipe(
                        prompt=prompt,
                        image=original_image,
                        mask_image=mask_image,
                        num_inference_steps=inpaint_step,
                        guidance_scale=inpaint_gs
                    ).images[0]
                elapsed = time.time() - start

                results.append(((ctrl_gs, ctrl_step), (inpaint_gs, inpaint_step), elapsed))
                images.append((inpainted_result, ctrl_gs, ctrl_step, inpaint_gs, inpaint_step, elapsed))

# === Display grid ===
cols = len(inpaint_steps) * len(inpaint_guidance_scales)
rows = len(ctrl_steps) * len(ctrl_guidance_scales)
fig, axs = plt.subplots(rows, cols, figsize=(3 * cols, 3 * rows))

for i, (img, cgs, cst, igs, ist, t) in enumerate(images):
    row = i // cols
    col = i % cols
    axs[row, col].imshow(img)
    axs[row, col].set_title(f"C[GS={cgs},S={cst}]\nI[GS={igs},S={ist}]\n{t:.2f}s", fontsize=7)
    axs[row, col].axis("off")

plt.tight_layout()
plt.show()

# === Summary ===
print("\n=== Grid Search Summary ===")
for (cgs, cst), (igs, ist), t in results:
    print(f"ControlNet: GS={cgs}, Steps={cst} | Inpaint: GS={igs}, Steps={ist} → {t:.2f}s")

fastest = min(results, key=lambda x: x[2])
print(f"\n✅ Fastest Combo: C[GS={fastest[0][0]}, S={fastest[0][1]}] + I[GS={fastest[1][0]}, S={fastest[1][1]}] → {fastest[2]:.2f}s")
```
