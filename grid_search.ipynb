```python
import os
import time
import numpy as np
from PIL import Image, ImageEnhance, ImageOps
import torch
import matplotlib.pyplot as plt
from diffusers import StableDiffusionControlNetPipeline, ControlNetModel
from diffusers.utils import load_image
from tqdm import tqdm

# Set specific GPU
os.environ["CUDA_VISIBLE_DEVICES"] = "1"

# File paths and prompt
# original_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/standalone_shirts/1907.jpg"
# sketch_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/sketches/1907_sketch.png"
original_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/standalone_shirts/12583.jpg"
sketch_path = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/sketches/12583_sketch.png"
prompt = "pink sweater"

# Load and preprocess images
original_image = load_image(original_path).resize((512, 512)).convert("RGB")
sketch_image = Image.open(sketch_path).resize((512, 512)).convert("L")
sketch_image = ImageOps.invert(sketch_image)
sketch_image = ImageEnhance.Contrast(sketch_image).enhance(2.0).convert("RGB")

# Load ControlNet model
controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/control_v11p_sd15_scribble",
    torch_dtype=torch.float16
).to("cuda")

pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    controlnet=controlnet,
    torch_dtype=torch.float16
).to("cuda")

pipe.safety_checker = None
pipe.enable_attention_slicing()

# Grid search configuration
guidance_scales = list(np.arange(4.0, 10.5, 0.5))  # 4.0 to 10.0 inclusive

inference_steps = list(range(20, 65, 5))  # 20, 25, ..., 60

# guidance_scales = list(np.arange(4.0, 10.5, 3))  # 4.0 to 10.0 inclusive

# inference_steps = list(range(20, 65, 15))  # 20, 25, ..., 60


results = []
images = []

# Run generation
for gs in guidance_scales:
    for steps in inference_steps:
        print(f"Running: guidance_scale={gs}, steps={steps}")
        start = time.time()

        with torch.inference_mode(), torch.autocast("cuda"):
            image = pipe(
                prompt=prompt,
                image=sketch_image,
                num_inference_steps=steps,
                guidance_scale=gs,
                controlnet_conditioning_scale=1.5
            ).images[0]

        elapsed = time.time() - start
        results.append((gs, steps, elapsed))
        images.append((image, gs, steps, elapsed))

# Display all outputs in a big grid
cols = len(inference_steps)
rows = len(guidance_scales)
fig, axs = plt.subplots(rows, cols, figsize=(3 * cols, 3 * rows))

for i, (img, gs, steps, t) in enumerate(images):
    row = i // cols
    col = i % cols
    axs[row, col].imshow(img)
    axs[row, col].set_title(f"GS={gs}, Steps={steps}\n{t:.2f}s", fontsize=8)
    axs[row, col].axis("off")

plt.tight_layout()
plt.show()

# Print timing summary
print("\n=== Inference Time Summary ===")
for gs, steps, t in results:
    print(f"GS={gs}, Steps={steps} ⟶ {t:.2f}s")

fastest = min(results, key=lambda x: x[2])
print(f"\n✅ Fastest: GS={fastest[0]}, Steps={fastest[1]}, Time={fastest[2]:.2f}s")
```

```python
import torch
from PIL import Image
from torchvision import transforms
from transformers import CLIPProcessor, CLIPModel

# Load CLIP model (ViT-B/32 by default)
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").eval().cuda()
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

# Function to compute CLIP score between an image and a prompt
def compute_clip_score(image_path, prompt):
    image = Image.open(image_path).convert("RGB")

    inputs = processor(text=[prompt], images=image, return_tensors="pt", padding=True).to("cuda")
    with torch.no_grad():
        outputs = model(**inputs)
        logits_per_image = outputs.logits_per_image  # cosine similarity
        score = logits_per_image.item()

    return score
prompt = "blue shirt"
image_path = "blue_shirt.png"  # path to your output image
clip_score = compute_clip_score(image_path, prompt)
print(f"CLIP score: {clip_score:.4f}")
```
