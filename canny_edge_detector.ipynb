```python
import os
import cv2
from PIL import Image
import numpy as np

# 📁 Paths
input_folder = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/standalone_bottomwears"       # folder with original T-shirt images
output_folder = "/data/home/huixian/Documents/677_project/fashion/fashion-dataset/sketches_bottomwears"    # folder where sketches will be saved

os.makedirs(output_folder, exist_ok=True)

# 🔍 Parameters for edge detection
low_threshold = 100
high_threshold = 200

# 🖼️ Loop through all images and generate sketches
for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        # Load image in grayscale for edge detection
        image_path = os.path.join(input_folder, filename)
        image = cv2.imread(image_path)
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        # Apply Gaussian Blur (optional, reduces noise)
        blurred = cv2.GaussianBlur(gray, (5, 5), 0)

        # 🧵 Apply Canny edge detection
        edges = cv2.Canny(blurred, low_threshold, high_threshold)

        # Invert for scribble-style (black on white)
        sketch = 255 - edges

        # Save as PNG
        out_path = os.path.join(output_folder, os.path.splitext(filename)[0] + "_sketch.png")
        cv2.imwrite(out_path, sketch)

        print(f"Saved sketch: {out_path}")
```
